## QUESTION 01: WHAT IS OBJECT-ORIENTED PROGRAMMING (OOP)?

Definition:
Object-Oriented Programming (OOP) is a programming paradigm
that organizes a program using classes and objects.
It combines data (attributes) and behaviour (methods) into objects.

Real-life example:
A car has attributes such as brand, color, and model.
It has behaviours such as start(), stop(), and accelerate().

Advantages of OOP:
1. Code reusability: Existing code can be reused.
2. Modularity: A program can be divided into classes.
3. Maintainability: Code is easier to update and maintain.
4. Data protection: Encapsulation helps control access to data.
5. Flexibility: Inheritance and polymorphism support flexible designs.

Disadvantages of OOP:
1. Learning curve: Classes and objects can be difficult for beginners.
2. More code: Small programs may need more code than procedural programs.
3. Memory overhead: Objects can require additional memory.
4. Design complexity: Poor class design can make a program complicated.

Python Example:
Expected output:
Toyota car is starting.

````python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def start(self):
        print(self.brand, "car is starting.")


car1 = Car("Toyota", "White")
car1.start()
````

## QUESTION 02: WHAT IS FUNCTIONAL PROGRAMMING?

Definition:
Functional Programming (FP) is a programming paradigm that
builds programs by using and combining functions.
It emphasizes pure functions, avoiding unnecessary side effects,
and transforming data rather than changing shared state.

Important concepts:
1. Pure function: Same input gives the same output, with no side effects.
2. Higher-order function: Accepts a function or returns a function.
3. Lambda: A short anonymous function.
4. Immutability: Avoid changing existing data.
5. map(), filter(), and reduce(): Common functional tools.

Python Example:
Expected output:
Squares: [1, 4, 9, 16, 25]
Even numbers: [2, 4]
Sum: 15

````python
from functools import reduce

def square(number):
    return number * number


numbers = [1, 2, 3, 4, 5]

squares = list(map(square, numbers))
even_numbers = list(filter(lambda number: number % 2 == 0, numbers))
total = reduce(lambda first, second: first + second, numbers)

print("Squares:", squares)
print("Even numbers:", even_numbers)
print("Sum:", total)
````

## QUESTION 03: DIFFERENCE BETWEEN FUNCTIONAL PROGRAMMING AND OOP

Functional Programming:
- Organizes program logic around functions.
- Focuses on transforming data using functions.
- Usually avoids changing existing data and shared state.
- Pure functions are easier to test independently.

Object-Oriented Programming:
- Organizes code around classes and objects.
- Combines data and behaviour in objects.
- Objects can hold and modify their own state.
- Uses concepts such as encapsulation, inheritance,
  polymorphism, and abstraction.

Comparison:
Feature                  Functional Programming       OOP
Main building block      Functions                    Classes and objects
Main focus               Data transformation          Objects and their behaviour
State                    Avoids shared mutable state   Objects may maintain state
Reusability              Function composition          Classes, inheritance, composition
Common concepts          Pure functions, map, filter   Encapsulation, inheritance

Python supports both programming paradigms.

Functional-style example:
OOP-style example:
Expected output:

OOP style: 25

````python
def calculate_square(number):
    return number ** 2


print("Functional style:", calculate_square(5))



class Calculator:
    def square(self, number):
        return number ** 2


calculator = Calculator()
print("OOP style:", calculator.square(5))
````

## QUESTION 15: WHAT IS ENCAPSULATION?

Definition:
Encapsulation is an OOP concept that bundles data (attributes)
and methods into a single unit (class), and controls how the
internal data is accessed or modified.

In Python, naming conventions and properties are commonly used
to manage access to attributes.

Benefits:
1. Helps protect an object's internal state.
2. Allows validation before changing data.
3. Improves maintainability.
4. Hides implementation details behind methods or properties.

Python Example:
Expected output:
Deposit successful.
Balance: 1500

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit successful.")
        else:
            print("Deposit amount must be positive.")

    def get_balance(self):
        return self.__balance


account = BankAccount("Ankan", 1000)
account.deposit(500)
print("Balance:", account.get_balance())
````

## QUESTION 17: WHAT ARE GETTERS AND SETTERS?

Definition:
A getter is a method or property used to read an attribute.
A setter is a method or property used to validate and update
an attribute.

In Python, @property provides a convenient getter and
@attribute.setter provides a setter.

Python Example:
Getter
Setter with validation
Expected output:
Current age: 22
Updated age: 23

````python
class Person:
    def __init__(self, age):
        self.__age = age

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value >= 0:
            self.__age = value
        else:
            raise ValueError("Age cannot be negative.")


person = Person(22)

print("Current age:", person.age)  # Calls the getter

person.age = 23                    # Calls the setter
print("Updated age:", person.age)
````

## QUESTION 18: WHAT IS ABSTRACTION? WHAT IS AN ABSTRACT BASE CLASS?

Definition of Abstraction:
Abstraction is an OOP concept that hides implementation details
and exposes only the essential features of an object.

Abstract Base Class (ABC):
An abstract base class defines a common interface for its
subclasses. It can contain abstract methods that subclasses
are required to implement.

Python provides ABC and abstractmethod through the abc module.

Python Example:
Expected output:
Area: 50

Note:
We cannot normally create an object directly from Shape because
it contains an unimplemented abstract method.

````python
from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


rectangle = Rectangle(10, 5)
print("Area:", rectangle.area())
````

## QUESTION 20: WHAT ARE THE TYPES OF INHERITANCE IN PYTHON?

**Definition:** Inheritance is an Object-Oriented Programming feature in which a child class acquires attributes and methods from a parent class. It supports code reusability and allows related classes to share common behaviour.

Python commonly demonstrates five inheritance structures:

1. Single Inheritance
2. Multiple Inheritance
3. Multilevel Inheritance
4. Hierarchical Inheritance
5. Hybrid Inheritance

Each type is explained separately below.

### 20.1 SINGLE INHERITANCE

**Definition:** Single inheritance occurs when one child class inherits from one parent class.

**Structure:** One parent → One child.

**Example:** A `Car` class inherits common features from a `Vehicle` class. The car can use the parent's `start()` method and also define its own `drive()` method.

````python
class Vehicle:
    def start(self):
        print("Vehicle is starting.")


class Car(Vehicle):
    def drive(self):
        print("Car is moving.")


my_car = Car()
my_car.start()
my_car.drive()
````

**Expected Output:**
```text
Vehicle is starting.
Car is moving.
```

**Explanation:** `Car` inherits the `start()` method from `Vehicle`. It also has its own `drive()` method.

### 20.2 MULTIPLE INHERITANCE

**Definition:** Multiple inheritance occurs when one child class inherits from two or more parent classes.

**Structure:** Parent 1 + Parent 2 → One child.

**Example:** A `SmartPhone` class inherits calling features from `Phone` and internet features from `Computer`.

````python
class Phone:
    def make_call(self):
        print("Making a phone call.")


class Computer:
    def browse_internet(self):
        print("Browsing the internet.")


class SmartPhone(Phone, Computer):
    pass


device = SmartPhone()
device.make_call()
device.browse_internet()
````

**Expected Output:**
```text
Making a phone call.
Browsing the internet.
```

**Explanation:** `SmartPhone` can use methods from both parent classes, `Phone` and `Computer`.

**Important:** When parent classes contain methods with the same name, Python uses Method Resolution Order (MRO) to decide which method is found first. You can inspect it using `SmartPhone.mro()`.

### 20.3 MULTILEVEL INHERITANCE

**Definition:** Multilevel inheritance occurs when a class inherits from another child class, creating an inheritance chain.

**Structure:** Grandparent → Parent → Child.

**Example:** ` puppy ` inherits from `Dog`, and `Dog` inherits from `Animal`. The puppy can access methods defined in both `Dog` and `Animal`.

````python
class Animal:
    def eat(self):
        print("Animal eats food.")


class Dog(Animal):
    def bark(self):
        print("Dog barks.")


class Puppy(Dog):
    def weep(self):
        print("Puppy makes a soft sound.")


my_puppy = Puppy()
my_puppy.eat()
my_puppy.bark()
my_puppy.weep()
````

**Expected Output:**
```text
Animal eats food.
Dog barks.
Puppy makes a soft sound.
```

**Explanation:** `Puppy` inherits `bark()` from `Dog` and `eat()` from `Animal` through the inheritance chain.

### 20.4 HIERARCHICAL INHERITANCE

**Definition:** Hierarchical inheritance occurs when two or more child classes inherit from the same parent class.

**Structure:** One parent → Multiple child classes.

**Example:** `Dog` and `Cat` both inherit the common `Animal` class. Each child class also has its own behaviour.

````python
class Animal:
    def eat(self):
        print("Animal eats food.")


class Dog(Animal):
    def bark(self):
        print("Dog barks.")


class Cat(Animal):
    def meow(self):
        print("Cat meows.")


dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
````

**Expected Output:**
```text
Animal eats food.
Dog barks.
Animal eats food.
Cat meows.
```

**Explanation:** Both `Dog` and `Cat` inherit the `eat()` method from `Animal`, but each class has its own additional method.

### 20.5 HYBRID INHERITANCE

**Definition:** Hybrid inheritance is a combination of two or more types of inheritance, such as hierarchical and multiple inheritance.

**Structure:** The exact structure depends on the inheritance types being combined.

**Example:** `Student` and `Employee` inherit from `Person`. The `TeachingAssistant` class inherits from both `Student` and `Employee`. This combines hierarchical and multiple inheritance.

````python
class Person:
    def introduce(self):
        print("I am a person.")


class Student(Person):
    def study(self):
        print("I study.")


class Employee(Person):
    def work(self):
        print("I work.")


class TeachingAssistant(Student, Employee):
    def assist(self):
        print("I assist the teacher.")


ta = TeachingAssistant()
ta.introduce()
ta.study()
ta.work()
ta.assist()
````

**Expected Output:**
```text
I am a person.
I study.
I work.
I assist the teacher.
```

**Explanation:** `TeachingAssistant` inherits from both `Student` and `Employee`. Both of those classes inherit from `Person`. Python's MRO handles the shared base class consistently.

**Quick Revision – Five Types of Inheritance**

| Type | Meaning |
|---|---|
| Single | One child inherits from one parent |
| Multiple | One child inherits from multiple parents |
| Multilevel | Inheritance takes place in a chain |
| Hierarchical | Multiple children inherit from one parent |
| Hybrid | Combination of two or more inheritance types |

## QUESTION 21: WHAT IS THE super() KEYWORD?

Definition:
super() returns a proxy that delegates method calls to the
next class in the Method Resolution Order (MRO).
It is commonly used in a child class to call a parent class
method, including the parent's initializer.

Python Example:
Expected output:
Name: Ankan
Course: MCA

````python
class Person:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name:", self.name)


class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def display(self):
        super().display()
        print("Course:", self.course)


student = Student("Ankan", "MCA")
student.display()
````

## QUESTION 22: WHAT IS POLYMORPHISM? WHAT ARE ITS TYPES IN PYTHON?

**Definition:** Polymorphism means “many forms”. In OOP, it allows the same interface, method name, or operator to perform different actions depending on the object or data involved.

**Why is polymorphism useful?**
1. It allows a common interface for different object types.
2. It makes code more flexible and extensible.
3. It reduces the need for repeated conditional logic.
4. It supports reusable and maintainable programs.

The following are common forms of polymorphism demonstrated in Python. Each is explained separately.

### 22.1 DUCK TYPING

**Definition:** Duck typing is a Python programming approach in which an object's suitability is determined by the methods or behaviour it provides, rather than by its exact class.

The idea is: if an object provides the required behaviour, it can be used.

**Example:** Different objects have a `sound()` method. A function can call `sound()` without checking the object's class.

````python
class Duck:
    def sound(self):
        print("Quack")


class Person:
    def sound(self):
        print("Hello")


def make_sound(obj):
    obj.sound()


make_sound(Duck())
make_sound(Person())
````

**Expected Output:**
```text
Quack
Hello
```

**Explanation:** `make_sound()` works with both objects because both provide the required `sound()` method.

### 22.2 METHOD OVERRIDING (RUNTIME POLYMORPHISM)

**Definition:** Method overriding occurs when a child class provides its own implementation of a method already defined in its parent class.

When the overridden method is called on a child object, Python executes the child class's implementation.

**Example:** `Dog` and `Cat` override the `sound()` method inherited from `Animal`.

````python
class Animal:
    def sound(self):
        print("Animal makes a sound.")


class Dog(Animal):
    def sound(self):
        print("Dog barks.")


class Cat(Animal):
    def sound(self):
        print("Cat meows.")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

**Expected Output:**
```text
Dog barks.
Cat meows.
```

**Explanation:** Both child classes have a method named `sound()`, but each provides a different implementation.

### 22.3 OPERATOR OVERLOADING

**Definition:** Operator overloading allows operators such as `+`, `-`, and `*` to work with user-defined objects by implementing special methods.

For example, the `+` operator can be customized using the `__add__()` method.

**Example:** Add two objects of a user-defined `Number` class.

````python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


number1 = Number(10)
number2 = Number(20)

result = number1 + number2
print("Sum:", result.value)
````

**Expected Output:**
```text
Sum: 30
```

**Explanation:** The expression `number1 + number2` internally calls `number1.__add__(number2)`. The method returns a new `Number` object containing the sum.

### 22.4 METHOD OVERLOADING-LIKE BEHAVIOUR

**Definition:** Traditional method overloading means defining multiple methods with the same name but different parameter lists. Python does not support this form of method overloading directly: if the same method is defined multiple times in a class, the later definition replaces the earlier one.

However, similar flexibility can be achieved using default arguments, `*args`, or other argument-handling techniques.

**Example:** One `add()` method accepts one, two, or three numbers using default parameter values.

````python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c


calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
````

**Expected Output:**
```text
10
30
60
```

**Explanation:** The same method works with different numbers of arguments because `b` and `c` have default values.

**Important Interview Point:** This is method overloading-like behaviour, not traditional signature-based method overloading.

### POLYMORPHISM – QUICK REVISION

| Operator overloading | Customizes operators using special methods |

## QUESTION 23: WHAT ARE SHALLOW COPY AND DEEP COPY?

**Definition of Copy:** Copying means creating another object based on an existing object.

Python provides the `copy` module to create shallow and deep copies.

- `copy.copy()` creates a shallow copy.
- `copy.deepcopy()` creates a deep copy.

These concepts are especially important when an object contains nested mutable objects, such as a list containing other lists.

### 23.1 SHALLOW COPY

**Definition:** A shallow copy creates a new outer object, but it does not recursively copy the objects contained inside it. References to nested objects are shared between the original and the copied object.

**Key points:**
1. A new outer object is created.
2. Nested objects are shared.
3. A change to a shared nested mutable object can appear in both objects.
4. Use `copy.copy()` to create a shallow copy.

**Example:** Create a shallow copy of a nested list and change an item inside a nested list.

````python
import copy

original_list = [[1, 2], [3, 4]]

shallow_copy_list = copy.copy(original_list)

shallow_copy_list[0][0] = 100

print("Original list:", original_list)
print("Shallow copy:", shallow_copy_list)
````

**Expected Output:**
```text
Original list: [[100, 2], [3, 4]]
Shallow copy: [[100, 2], [3, 4]]
```

**Explanation:** The outer lists are different objects, but the nested list `[1, 2]` is shared. Therefore, changing that nested list through the copied object also affects the original.

### 23.2 DEEP COPY

**Definition:** A deep copy creates a new outer object and recursively copies the objects contained inside it. Nested mutable objects are therefore independent of the original.

**Key points:**
1. A new outer object is created.
2. Nested objects are recursively copied.
3. Changes to a copied nested mutable object do not affect the original.
4. Use `copy.deepcopy()` to create a deep copy.

**Example:** Create a deep copy of a nested list and change an item inside the copied nested list.

````python
import copy

original_list = [[1, 2], [3, 4]]

deep_copy_list = copy.deepcopy(original_list)

deep_copy_list[0][0] = 100

print("Original list:", original_list)
print("Deep copy:", deep_copy_list)
````

**Expected Output:**
```text
Original list: [[1, 2], [3, 4]]
Deep copy: [[100, 2], [3, 4]]
```

**Explanation:** The nested lists are copied recursively. Therefore, changing the nested list inside the deep copy does not change the original list.

### 23.3 DIFFERENCE BETWEEN SHALLOW COPY AND DEEP COPY

| Feature | Shallow Copy | Deep Copy |
|---|---|---|
| Outer object | New object | New object |
| Nested objects | References are shared | Recursively copied |
| Nested changes | May affect the original | Do not affect the original for copied mutable objects |
| Function | `copy.copy()` | `copy.deepcopy()` |
| Typical use | Flat collections or when sharing nested objects is acceptable | Nested data that needs independent copies |

**Important Note:** Deep copy is not always necessary. Choose it when nested mutable objects must be independent. Copying custom objects can also involve special behaviour, and some resources cannot be copied normally.
